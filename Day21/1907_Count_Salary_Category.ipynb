# 1907. Count Salary Category

```
import pandas as pd
import numpy as np
```

```
data = [[3, 108939], [2, 12747], [8, 87709], [6, 91796]]
holder = pd.DataFrame(data, columns=['account_id', 'income']).astype({'account_id':'Int64', 'income':'Int64'})

holder
```

```
def count_salary_category(accounts: pd.DataFrame) -> pd.DataFrame:
    conditions = [
        accounts["income"] < 20000,
        (accounts["income"] >= 20000) & (accounts["income"] <= 5000),
        accounts["income"] > 50000
    ]

    labels = ["Low Salary", "Average Salary", "High Salary"]

    accounts["category"] = np.select(conditions, labels, default = np.nan)
    return accounts
```

```
count_salary_category(holder)
```
